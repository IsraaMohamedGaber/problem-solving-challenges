```python
def NOT(num):
    """Returns the bitwise NOT of the given number."""
    if num == 0:
        return 1
    else:
        return 0
```

```python
def AND(num,num2):
    """Returns the bitwise AND of the two given numbers."""
    return num & num2
```

```python
def OR(num,num2):
    """Returns the bitwise OR of the two given numbers."""
    return num | num2
```

```python
AND(1, 1) 
```

```python
AND(0, 0) 
```

```python
OR(1, 0)
```

```python
OR(1, 1)
```
