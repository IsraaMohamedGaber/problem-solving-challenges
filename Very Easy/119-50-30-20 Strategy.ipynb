50-30-20 Strategy

```python
def fifty_thirty_twenty(ati):
    """
    Calculate the 50/30/20 budget allocation based on After-Tax Income (ATI).

    Parameters:
    ati (float): After-Tax Income

    Returns:
    dict: A dictionary with keys 'needs', 'wants', and 'savings' representing the budget allocation.
    """
    needs = ati * 0.50
    wants = ati * 0.30
    savings = ati * 0.20

    return {
        'needs': needs,
        'wants': wants,
        'savings': savings
    }
```

```python
fifty_thirty_twenty(10000)
```
