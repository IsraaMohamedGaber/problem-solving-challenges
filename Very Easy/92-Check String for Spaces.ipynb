Check String for Spaces


Problem Source

https://edabit.com/challenge/rR2qf7ELnXoXESiz2

```python
def has_spaces(txt):
    if ' ' in txt:
        return True
    else:     
        return False
```

```python
has_spaces("hello")
```

```python
has_spaces("hello, world") 
```

```python
has_spaces(" ") 
```

```python
has_spaces("")
```

```python
has_spaces(",./!@#")
```
