Slightly Superior


Problem Source

https://edabit.com/challenge/ZF6vZwPc5He5u5EFe

```python
def is_first_superior(lst1, lst2):
    return lst1[0] > lst2[0]    
```

```python
is_first_superior([1, 2, 4], [1, 2, 3]) 
```

```python
is_first_superior(["a", "d", "c"], ["a", "b", "c"]) 
```

```python
is_first_superior(["zebra", "ostrich", "whale"], ["ant", "ostrich", "whale"])
```

```python
is_first_superior([1, 2, 3, 4], [1, 2, 4, 4]) 
```

```python
is_first_superior([True, 10, "zebra"], [True, 10, "zebra"]) 
```
