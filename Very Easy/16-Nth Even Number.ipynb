Nth Even Number


Problem Source

https://edabit.com/challenge/WYq4aFwSNuoyFCW5G

```python
#Create a function that takes a number n and returns the nth even number beginning with 0 as the first.

def nth_even(n):
    return (n - 1) * 2
```

```python
nth_even(1)
```

```python
nth_even(2)
```

```python
nth_even(100)
```
