The 3 Programmers Problem


Problem Source

https://edabit.com/challenge/9zsDKijmBffmnk9AP

```python
def programmers(one, two, three):
    """Return the number of programmers based on their counts."""
    return max(one, two, three) - min(one, two, three)
```

```python
programmers(147, 33, 526)
```

```python
programmers(33, 72, 74) 
```

```python
programmers(1, 5, 9)
```
