Check If It's a Title String


Problem Source

https://edabit.com/challenge/CHwX2o6rqrBsL4gzr

```python
def check_title(txt):
    if txt.istitle():
        return True
    return False
```

```python
check_title("A Mind Boggling Achievement")
```

```python
check_title("A Simple Python Program!")
```

```python
check_title("Water is transparent") 
```
