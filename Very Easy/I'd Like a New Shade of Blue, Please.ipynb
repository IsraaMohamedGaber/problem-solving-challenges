```python
def how_many_walls(n, w, h):
    """Calculates how many walls of dimensions w x h can be painted with n liters of paint."""
    wall_area = w * h
    if wall_area == 0:
        return 0  # Avoid division by zero
    return n // wall_area
```

```python
how_many_walls(100, 4, 5)
```

```python
how_many_walls(10, 15, 12)
```

```python
how_many_walls(41, 3, 6) 
```
