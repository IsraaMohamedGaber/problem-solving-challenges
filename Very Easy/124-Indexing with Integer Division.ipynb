Indexing with Integer Division


Problem Source

https://edabit.com/challenge/FtwouqMohuumDZh23

```python
#Given an index using INTEGER division and a list, return the value of the list with the given index.
def value_at(lst, index):
    return lst[int(index)]   
```

```python
value_at([1, 2, 3, 4, 5, 6], 10 // 2) 
```

```python
value_at([1, 2, 3, 4, 5, 6], 8.0 // 2) 
```

```python
value_at([1, 2, 3, 4], 6.535355314 // 2) 
```
