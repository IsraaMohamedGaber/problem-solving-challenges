Return the First and Last Elements in a List


Problem Source

https://edabit.com/challenge/9kkSyfN62E4At9wcy

```python
def first_last(lst):
    return [lst[0], lst[-1]]
```

```python
first_last([5, 10, 15, 20, 25])
```

```python
first_last(["edabit", 13, None, False, True])
```

```python
first_last([None, 4, "6", "hello", None])
```
