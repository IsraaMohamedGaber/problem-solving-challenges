Make My Way Home


Problem Source

https://edabit.com/challenge/hmvH3d8E2pKhCjsu3

```python
def distance_home(lst):
    final_position = sum(lst)
    return abs(final_position)
```

```python
distance_home([2, 4, 2, 5]) 
```

```python
distance_home([-1, -4, -3, -2])
```

```python
distance_home([3, 4, -5, -2]) 
```
