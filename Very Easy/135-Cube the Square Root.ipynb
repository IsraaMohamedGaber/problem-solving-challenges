```python
#Create a function that takes a number as an argument and returns the square root of that number cubed.
import math
def cube_squareroot(num):
    return int(math.sqrt(num ** 3))   
```

```python
cube_squareroot(81)
```

```python
cube_squareroot(1646089)
```
