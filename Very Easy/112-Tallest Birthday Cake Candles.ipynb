Tallest Birthday Cake Candles


Problem Source

https://edabit.com/challenge/PtrPzWCWrPW54xfxK

```python
def birthday_cake_candles(candles):
    tallest = max(candles)
    return candles.count(tallest)
```

```python
birthday_cake_candles([4, 4, 1, 3])
```

```python
birthday_cake_candles([3, 2, 1, 3])
```

```python
birthday_cake_candles([82, 49, 82, 82, 41, 82, 15, 63, 38, 25]) 
```
