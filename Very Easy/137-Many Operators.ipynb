Many Operators!


Problem Source

https://edabit.com/challenge/r3T23PTQnZRgjWGFH

```python
def operate(num1, num2, operator):
    if operator == '+':
        return num1 + num2
    elif operator == '-':
        return num1 - num2
    elif operator == '*':
        return num1 * num2
    elif operator == '%':
        return num1 % num2
    elif operator == '/':
        if num2 == 0:
            return "Error: Division by zero"
        return num1 / num2
    else:
        return "Error: Invalid operator"
```

```python
operate(1, 2, "+")
```

```python
operate(7, 10, "-") 
```

```python
operate(20, 10, "%")
```
