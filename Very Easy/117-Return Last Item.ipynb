Return Last Item


Problem Source

https://edabit.com/challenge/mcC546MLnBjNLXTb8

```python
def last_ind(lst):
    if lst == []:
        return None 
    return lst[-1]
```

```python
last_ind([0, 4, 19, 34, 50, -9, 2])
```

```python
last_ind("The quick brown fox jumped over the lazy dog") 
```

```python
last_ind([]) 
```
