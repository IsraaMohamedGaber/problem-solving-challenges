```python
def create_id(firstname, lastname):
    id_part1 = firstname[0].lower()
    id_part2 = lastname[0].upper()
    id_part3 =lastname[1:3].lower()
    return id_part1 + id_part2  + id_part3
```

```python
create_id("mary", "lamb")
```

```python
create_id("John", "SMITH")
```
