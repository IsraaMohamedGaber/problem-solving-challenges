Half, Quarter and Eighth


Problem Source

https://edabit.com/challenge/u68XgCZcWGphs5R54

```python
def half_quarter_eighth(n):
    half = n / 2
    quarter = n / 4
    eighth = n / 8
    return half, quarter, eighth
```

```python
half_quarter_eighth(6)
```

```python
half_quarter_eighth(22)
```

```python
half_quarter_eighth(25)
```
