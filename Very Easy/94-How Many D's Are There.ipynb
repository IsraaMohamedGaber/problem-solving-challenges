How Many D's Are There?


Problem Source

https://edabit.com/challenge/xdSKkXQkkMroNzq8C

```python
def count_d(sentence):
    return sentence.lower().count('d')  
```

```python
count_d("My friend Dylan got distracted in school.") 
```

```python
count_d("Debris was scattered all over the yard.") 
```

```python
count_d("The rodents hibernated in their den.") 
```
