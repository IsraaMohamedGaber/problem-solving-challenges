Slice of Pie


Problem Source

https://edabit.com/challenge/fqn5FcLzEb4RBH9w7

```python
def equal_slices(total, people, each):
    if people == 0:
        return False
    return total >= people * each
```

```python
equal_slices(11, 5, 2)
```

```python
equal_slices(11, 5, 3)
```

```python
equal_slices(8, 3, 2)
```

```python
equal_slices(8, 3, 3) 
```

```python
equal_slices(24, 12, 2) 
```
