Nothing is Nothing?


Problem Source

https://edabit.com/challenge/n26hBhqdRk9ToTboA

```python
def nothing_is_nothing(*args):
    for arg in args:
        if not arg:
            return False
    return True 
```

```python
nothing_is_nothing(0, False, [], {}) 
```

```python
nothing_is_nothing(33, "Hello", (True, True, 3)) 
```

```python
nothing_is_nothing(True, None) 
```
