Instant JAZZ


Problem Source

https://edabit.com/challenge/jhghtvT2s58FnDr5T

```python
def jazzify(lst):
    return [s if s.endswith('7') else s + '7' for s in lst]
```

```python
jazzify(["G", "F", "C"])
```

```python
jazzify(["Dm", "G", "E", "A"])
```

```python
jazzify(["F7", "E7", "A7", "Ab7", "Gm7", "C7"])
```

```python
jazzify([])
```
