CMS Selector Based on a Given String


Problem Source

https://edabit.com/challenge/whEretKtpEbEBAAfr

The search should be case-insensitive (e.g., "w" matches "WordPress").


If the pattern is an empty string, return all CMS names, sorted alphabetically.

```python
def cms_selector(lst, txt):
    if txt == "":
        return sorted(lst)  
    else:
        pattern = txt.lower()
        result = [cms for cms in lst if txt in cms.lower()]
        return sorted(result)
```

```python
cms_selector(["WordPress", "Joomla", "Drupal"], "w")
```

```python
cms_selector(["WordPress", "Joomla", "Drupal", "Magento"], "ru")
```

```python
cms_selector(["WordPress", "Joomla", "Drupal", "Magento"], "") 
```
