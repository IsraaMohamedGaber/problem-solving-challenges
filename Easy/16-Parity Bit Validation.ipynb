Parity Bit Validation


Problem Source

https://edabit.com/challenge/DcXTvFs7Zvc3PDTzX

General Rule

To validate:

Extract the last bit (parity bit).

Count the number of 1s in the first 7 bits (data).

If:

count of 1s is even and parity bit = 0 → True

count of 1s is odd and parity bit = 1 → True

otherwise → False

```python
def validate_binary(b):
    data = b[:-1]       # first 7 bits
    parity = b[-1]      # last bit
    ones_count = data.count('1') # count number of 1s

    # Check validity
    if (ones_count % 2 == 0 and parity == '0') or (ones_count % 2 == 1 and parity == '1'):
        return True
    return False
```

```python
validate_binary("00101101")
```

```python
validate_binary("11000000")
```

```python
validate_binary("11000001")
```
