Vowel Replacer


Problem Source

https://edabit.com/challenge/Ggq8GtYPeHJQg4v7q

```python
def replace_vowels(txt, ch):
    vowels = "aeiouAEIOU"
    for vowel in vowels:
        txt = txt.replace(vowel, ch)
    return txt
```

```python
replace_vowels("the aardvark", "#")
```

```python
replace_vowels("minnie mouse", "?")
```

```python
replace_vowels("shakespeare", "*")
```
