How Many Vowels?


Problem Source

https://edabit.com/challenge/p88k8yHRPTMPt4bBo

```python
def count_vowels(txt):      
    vowels = "aeiouAEIOU"
    count = 0
    for char in txt:
        if char in vowels:
            count += 1
    return count
```

```python
count_vowels("Celebration")
```

```python
count_vowels("Palm")
```

```python
count_vowels("Prediction") 
```
