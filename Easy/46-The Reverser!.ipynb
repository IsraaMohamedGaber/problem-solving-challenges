The Reverser!


Problem Source

https://edabit.com/challenge/9hQogtkbZSSJ8tYsG

```python
def reverse(txt):
    """Returns The "Reverser" takes a string as input and returns that string in reverse order, with the opposite case."""
    return txt.swapcase()[::-1]
```

```python
reverse("Hello World")
```

```python
reverse("ReVeRsE")
```
