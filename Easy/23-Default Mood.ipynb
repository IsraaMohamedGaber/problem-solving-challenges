```python
def mood_today(mood):
    """Returns a message based on the user's mood.

    Args:
        mood (str): The user's current mood.

    Returns:
        str: A message corresponding to the mood.
    """
    if mood == "":
        return "Today, I am feeling neutral"
    else:
        return "Today, I am feeling "+mood
    
```

```python
mood_today("happy")
```

```python
mood_today("sad")
```

```python
mood_today("")
```
