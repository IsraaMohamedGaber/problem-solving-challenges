Error Messages


Problem Source

https://edabit.com/challenge/GP8Tywnn2gucEfSMf


```python
def error(n):
    switch = {
        1: "Check the fan: e1",
        2: "Emergency stop: e2",
        3: "Pump Error: e3",
        4: "c: e4",
        5: "Temperature Sensor Error: e5",
    }
    return switch.get(n, "Invalid")
```

```python
error(1)
```

```python
error(2)
```

```python
error(3)
```
