How Heavy Is It?


Problem Source

https://edabit.com/challenge/iP4ixkQffELyHvHi5

```python
import math
def weight(r, h):
    volume_cm3 = math.pi * (r ** 2) * h   # Step 1: volume in cm³
    volume_dm3 = volume_cm3 / 1000        # Step 2: convert to dm³
    mass_kg = volume_dm3                  # Step 3: 1 dm³ = 1 kg
    return round(mass_kg, 2)              # Step 4: round to 2 decimals
```

```python
weight(4, 10)
```

```python
weight(30, 60)
```

```python
weight(15, 10)
```
