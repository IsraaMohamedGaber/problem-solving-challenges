Xs and Os, Nobody Knows


Problem Source

https://edabit.com/challenge/JSJEuuWduBB5hEX6k

```python
def XO(txt):
    if txt.lower().count('x') == txt.lower().count('o'):
        return True
    else:
        return False
```

```python
XO("ooxx") 
```

```python
XO("xooxx")
```

```python
XO("ooxXm")
```

```python
XO("zpzpzpp")
```

```python
XO("zzoo")
```
