Hiding the Card Number


Problem Source

https://edabit.com/challenge/iRCwdDBkNcHM5QeAm

```python
def card_hide(card):
    """Hide all but the last four characters of a card number.

    Args:
        card (str): The card number as a string.

    Returns:
        str: The card number with all but the last four characters replaced by '*'.
    """
    if len(card) <= 4:
        return card
    return '*' * (len(card) - 4) + card[-4:]
```

```python
card_hide("1234123456785678")
```

```python
card_hide("8754456321113213")
```

```python
card_hide("35123413355523")
```
