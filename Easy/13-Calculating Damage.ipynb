Calculating Damage


Problem Source

https://edabit.com/challenge/HSHHkdRYXfgfZSqri

```python
def damage(damage, speed, time):
    if time == "second" :
        return damage * speed * 1
    elif time == "minute":
        return damage * speed * 60      
    else    : # time == "hour"
        return damage * speed * 3600
```

```python
damage(40, 5, "second")
```

```python
damage(100, 1, "minute")
```

```python
damage(2, 100, "hour")
```
