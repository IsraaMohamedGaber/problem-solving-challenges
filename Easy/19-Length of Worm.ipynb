Length of Worm


Problem Source

https://edabit.com/challenge/6LwzPRc6LrauEgr7H

```python
def worm_length(worm):
    if not worm or set(worm) != {'-'}:
        return "invalid"
    return f"{len(worm) * 10} mm."
```
