Moving to the End


Problem Source

https://edabit.com/challenge/nTW4KgmJxpLDXcWPt

```python
def move_to_end(lst, el):
    return [i for i in lst if i != el] + [i for i in lst if i == el]
```

```python
move_to_end([1, 3, 2, 4, 4, 1], 1)
```

```python
move_to_end([7, 8, 9, 1, 2, 3, 4], 9)
```

```python
move_to_end(["a", "a", "a", "b"], "a")
```
