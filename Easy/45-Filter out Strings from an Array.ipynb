Filter out Strings from an Array


Problem Source

https://edabit.com/challenge/nunJurLEibCyn8fzn

```python
def filter_list(lst):
	return [i for i in lst if type(i) == int]
```

```python
filter_list([1, 2, "a", "b"]) 
```

```python
filter_list([1, "a", "b", 0, 15])
```

```python

filter_list([1, 2, "aasf", "1", "123", 123])
```
