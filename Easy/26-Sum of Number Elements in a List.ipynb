Sum of Number Elements in a List


Problem Source

https://edabit.com/challenge/jQGT8CNFcMXr55jeb

```python
def numbers_sum(lst):
    total = 0
    for x in lst:
        if isinstance(x, (int, float)) and not isinstance(x, bool):
            total += x
    return total
    #return sum(x for x in lst if isinstance(x, (int, float)))
```

```python
numbers_sum([1, 2, "13", "4", "645"])
```

```python
numbers_sum([True, False, "123", "75"])
```

```python
numbers_sum([1, 2, 3, 4, 5, True])
```
