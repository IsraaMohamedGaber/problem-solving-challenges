Replace Vowel With Another Character


Problem Source

```python
# Create a function that takes a string and replaces the vowels with another character.

# a = 1
# e = 2
# i = 3
# o = 4
# u = 5
```

```python
def replace_vowel(word):
	new_word=word.lower()
	for char in new_word:
         if char == 'a':
            return new_word.replace(char, '1')
         elif char == 'e':   
            return new_word.replace(char, '2')
         elif char == 'i':
            return new_word.replace(char, '3')
         elif char == 'o':
            return new_word.replace(char, '4')  
         elif char == 'u':
            return new_word.replace(char, '5')
        
```

```python
replace_vowel("karachi")
```

```python
replace_vowel("chembur")
```

```python
replace_vowel("khandbari")
```
