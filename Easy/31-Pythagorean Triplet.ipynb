Pythagorean Triplet


Problem Source

https://edabit.com/challenge/Ns4Sjh7KK58ofAph8

```python
def is_triplet(n1, n2, n3):
    if n1 < n3 and n2 < n3:
        if n1**2 + n2**2 == n3**2:
            return True
        else:
            return False    
    elif n2 < n1 and n3 < n1:
        if n2**2 + n3**2 == n1**2:
            return True
        else:
            return False
    elif n3 < n2 and n1 < n2:
        if n1**2 + n3**2 == n2**2:
            return True
        else:
            return False    
```

```python
is_triplet(3, 4, 5)
```

```python
is_triplet(13, 5, 12)
```

```python
is_triplet(1, 2, 3)
```
