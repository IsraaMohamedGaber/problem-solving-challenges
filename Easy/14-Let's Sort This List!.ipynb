Problem Source

```python
def asc_des_none(lst, s):
    if s == "Asc":
        return sorted(lst)
    elif s == "Des":
        return sorted(lst, reverse=True)
    else:  # s == "None"
        return lst
```

```python
asc_des_none([4, 3, 2, 1], "Asc" )
```

```python
asc_des_none([7, 8, 11, 66], "Des") 
```

```python
asc_des_none([1, 2, 3, 4], "None")
```
