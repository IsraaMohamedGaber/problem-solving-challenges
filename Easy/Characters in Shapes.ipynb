```python
def count_characters(lst):
    """
    This function takes a list of strings and returns the total number of characters
    across all strings in the list.
    
    :param lst: List of strings
    :return: Total number of characters
    """
    total_characters = 0
    for string in lst:
        total_characters += len(string)
    return total_characters
```

```python
count_characters([
  "###",
  "###",
  "###"
])
```

```python
count_characters([
  "22222222",
  "22222222",
])
```

```python
count_characters([
  "------------------"
])
```

```python
count_characters([])
```

```python
count_characters(["", ""])
```
