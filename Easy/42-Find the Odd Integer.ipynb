Find the Odd Integer


Problem Source

https://edabit.com/challenge/9TcXrWEGH3DaCgPBs

```python
def find_odd(lst):
    for i in lst:
        if lst.count(i) % 2 != 0:
            return i
```

```python
find_odd([1, 1, 2, -2, 5, 2, 4, 4, -1, -2, 5]) 
```

```python
find_odd([20, 1, 1, 2, 2, 3, 3, 5, 5, 4, 20, 4, 5]) 
```

```python
find_odd([10])
```
