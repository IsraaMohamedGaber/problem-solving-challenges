Folding a Piece of Paper


Problem Source

https://edabit.com/challenge/4t6YAJS8dtT7RQjta

```python
#Create a function that returns the thickness (in meters) of a piece of paper after folding it n number of times. The paper starts off with a thickness of 0.5mm.
def num_layers(n):
    thickness = 0.0005  # initial thickness in meters
    for _ in range(n):
        thickness *= 2
    return thickness
```

```python
num_layers(1)
```

```python
num_layers(4)
```

```python
num_layers(21)
```
