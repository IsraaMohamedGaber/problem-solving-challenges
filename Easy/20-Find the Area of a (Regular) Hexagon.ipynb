Find the Area of a (Regular) Hexagon


Problem Source

https://edabit.com/challenge/LByfZDbfkYTyZs8cD

```python
import math
def area_of_hexagon(x):
    if not isinstance(x, (int, float)) or x <= 0:
        return None
# Step-by-Step Explanation
# 1️isinstance(x, (int, float))
# This checks if x is a number, specifically:
# an integer (like 1, 5, 10)
# or a float (like 1.5, 2.0, 3.7)
# Returns True if x is a number
# Returns False if x is something else (like a string, list, etc.)
# isinstance(3, (int, float))   # True
# isinstance(2.5, (int, float)) # True
# isinstance("5", (int, float)) # False
# isinstance([2], (int, float)) # False

    area = (3 * math.sqrt(3) / 2) * (x ** 2)
    return round(area, 1)
```

```python
area_of_hexagon(1)
```

```python
area_of_hexagon(2)
```
