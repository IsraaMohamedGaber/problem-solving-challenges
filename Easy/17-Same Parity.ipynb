Same Parity?


Problem Source

https://edabit.com/challenge/jzCGNwLpmrHQKmtyJ

```python
def parity_analysis(num):
    # Handle negative numbers by taking absolute value
    num = abs(num)
    
    # Calculate the sum of digits
    digit_sum = sum(int(digit) for digit in str(num))
    
    # Check if both the number and digit sum have the same parity
    return (num % 2) == (digit_sum % 2)
```

```python
parity_analysis(243) 
```

```python
parity_analysis(12) 
```

```python
parity_analysis(3)
```
