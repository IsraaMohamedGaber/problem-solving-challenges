Repeating Letters


Problem Source

https://edabit.com/challenge/HpqLxNqqRvMQoz8ME

```python
def double_char(txt):
    result = ""
    for char in txt:
        result += char * 2
    return result
```

```python
double_char("String")
```

```python
double_char("Hello World!")
```

```python
double_char("1234!_ ")
```
